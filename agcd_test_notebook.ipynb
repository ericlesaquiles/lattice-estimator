### Grid searching parameters for Cheon's scheme

```python
from estimator import AGCD
from estimator.reduction import RC
from math import log2
lam = 128
L = 1
for rho in range(100, 200, 1):
    eta = int(rho + L*log2(lam))
    for gamma in range(eta+10, 100*eta):        
        cheon_params =  AGCD.Parameters(gamma = gamma, eta = eta, rho = rho)
        result = AGCD.estimate(cheon_params, red_cost_model=RC.MATZOV, verbose=False)
        if(gamma%5000 == 0 and rho%2 == 0):
            print(L, result["lambda"], rho, gamma, eta)
        if result["lambda"] >= 128:
            print(L, result["lambda"], rho, gamma, eta)
            break
    if result["lambda"] >= 128:
        print(L, result["lambda"], rho, gamma, eta)
        break
```

```python
for L in range(2,10):
    print(f"L: {L}")
    for rho in range(90, 200, 2):
        eta = rho + L*7
        for gamma in range(eta+1, eta + 1000, 2):
            eta = rho + L*7
            cheon_params =  AGCD.Parameters(gamma = gamma, eta = eta, rho = rho)
            result = AGCD.estimate(cheon_params, red_cost_model=RC.MATZOV, verbose=False)
            #if gamma % 100 == 0:
            print(eta, gamma, rho, result["lambda"])
            if result["lambda"] > 128:
                print(L, result["lambda"], rho, gamma, eta)
                break
        if result["lambda"] > 128:
            result["lambda"] = 0
            break
```
