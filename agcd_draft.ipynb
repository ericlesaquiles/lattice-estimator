#### Orthogonal lattice attack estimation

1. get $\delta_0$ from each $\beta$
2. get $n$ corresponding to $\delta_0$ from the relation $\frac{\gamma - \rho}{n} - (\eta - \rho) +nlog\delta_0 + log \sqrt{n^2 + 2n} < 0$
3. compute the cost using the formula $c \cdot n \cdot 2^{0.292\beta + 16.4}$, where c is a small constant (we take it as 8)
